```python
# Import dependencies
import pandas as pd
```

```python
df = pd.read_excel(r'C:\Users\cameu\Desktop\Python Pandas Learning\Datasets\Customer Call List.xlsx')
df
```

```python
df.info()
```

```python
# Remove duplicates
df = df.drop_duplicates()
df
```

```python
# Remove unnecessary columns
df = df.drop(columns='Not_Useful_Column')
df
```

```python
# Remove special characters in Last Name column
df['Last_Name'] = df['Last_Name'].str.strip('123._/')
df
```

```python
df['Phone_Number'] = df['Phone_Number'].str.replace('[^a-zA-Z0-9]', '', regex=True)
df
```

```python

df['Phone_Number'] = df['Phone_Number'].apply(lambda x: str(x))
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: x[0:3] + '-' + x[3:6] + '-' + x[6:])
df
```

```python
df['Phone_Number'] = df['Phone_Number'].str.replace('nan--', '')
df['Phone_Number'] = df['Phone_Number'].str.replace('Na--', '')
df
```

```python
df[['Street Adress', 'State', 'Zip Code']] = df['Address'].str.split(',', n=2, expand=True)
df
```

```python
df['Paying Customer'] = df['Paying Customer'].replace('N', 'No')
df['Paying Customer'] = df['Paying Customer'].replace('Y', 'Yes')
df
```

```python
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace('N', 'No')
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace('Y', 'Yes')
df
```

```python
df = df.fillna('')
df
```

```python
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace('', 'No')
df
```

```python
for x in df.index:
    if df.loc[x, 'Do_Not_Contact'] == 'Yes':
        df.drop(x, inplace=True)

df
```

```python
df = df.reset_index(drop=True)
df
```
